```python
import os
from pathlib import Path

import numpy as np
from dotenv import load_dotenv
from openai import OpenAI

# Build an absolute path to .env from the repo root, no matter where we run from
env_path = Path.home() / "ai-engineering-lab" / "01_llm_api_logger" / ".env"
print("Looking for .env at:", env_path)
print("Does it exist?", env_path.exists())

load_dotenv(env_path)
print("OPENAI key loaded?", bool(os.getenv("OPENAI_API_KEY")))

client = OpenAI()
print("client ready")
```

```python
def embed(text):
    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=text,
    )
    return np.array(response.data[0].embedding)

v = embed("hello")
print(f"'hello' became a vector of {len(v)} numbers")
print(f"First 5 numbers: {v[:5]}")
```

```python
def cosine_similarity(a, b):
    # measures how closely two vectors point in the same direction: 1 = identical meaning, 0 = unrelated
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

sentences = [
    "I love my dog",
    "My puppy is wonderful",
    "The stock market crashed today",
]

# embed all three
vectors = [embed(s) for s in sentences]

# compare sentence 0 against the others
print(f"Base: {sentences[0]!r}\n")
for i in range(len(sentences)):
    sim = cosine_similarity(vectors[0], vectors[i])
    print(f"  vs {sentences[i]!r:35} -> similarity {sim:.3f}")
```
